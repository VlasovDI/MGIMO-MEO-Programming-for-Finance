# Seminar 3: Accessing Financial Data via APIs

Tasks using yfinance to download and analyze historical stock price data.

```python
#!pip install yfinance
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Task 1: Download stock prices data for Google (GOOGL) from January 1, 2020 to January 1, 2023.

Which method is used to download historical price data for a specific ticker object?

**Answer: `.history()`**

```python
stock = yf.Ticker("GOOGL")
data = stock.history(start="2020-01-01", end="2023-01-01", repair=True)
print(f"Downloaded {len(data)} trading days of GOOGL data")
data.head()
```

## Task 2: From May, 2020 to September, 2021 the price for Google stocks was mainly...

**Answer: growing**

```python
plt.figure(figsize=(10, 6))
plt.plot(data.index, data["Close"])
plt.title("GOOGL Stock Price (2020-2023)")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.grid(True, alpha=0.3)
plt.axvspan("2020-05-01", "2021-09-01", alpha=0.2, color="green", label="May 2020 - Sep 2021 (growing)")
plt.legend()
plt.show()
```

## Task 3: When was the peak price achieved in the period January 1, 2020 to January 1, 2023?

**Answer: December 2021**

```python
peak_date = data["Close"].idxmax()
peak_price = data["Close"].max()
print(f"Peak price: ${peak_price:.2f} on {peak_date.date()}")
```

## Task 4: What was the approximate mean simple return (January 1, 2020 to January 1, 2023)? Round to 4 decimal places.

**Answer: 0.0006**

```python
prices = data["Close"]
data["Simple_Return"] = prices.pct_change()
data = data.dropna()  # Remove first row with NaN
print(f"Mean Simple Return: {data['Simple_Return'].mean():.4f}")
```

## Task 5: When was the 30-Day Rolling Volatility (Annualized) of log return the highest for the period January 1, 2020 to January 1, 2023?

**Answer: April 2020**

```python
data["Log_Return"] = np.log(prices / prices.shift(1))
data = data.dropna()
data["Volatility"] = data["Log_Return"].rolling(window=30).std() * np.sqrt(252)

plt.figure(figsize=(10, 6))
plt.plot(data.index, data["Volatility"])
plt.title("30-Day Rolling Volatility (Annualized) - GOOGL")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.grid(True, alpha=0.3)
plt.show()

peak_vol_date = data["Volatility"].idxmax()
print(f"Highest volatility: {peak_vol_date.strftime('%B %Y')}")
```

## Task 6: Based on the stock prices data for Nike from January 1, 2022 to January 1, 2026, what was the peak price approximately equal to?

**Answer: ~$150**

```python
stock_nke = yf.Ticker("NKE")
data1 = stock_nke.history(start="2022-01-01", end="2026-01-01", repair=True)

plt.figure(figsize=(10, 6))
plt.plot(data1.index, data1["Close"])
plt.title("NKE Stock Price (2022-2025)")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.grid(True, alpha=0.3)
plt.show()

print(f"NKE Peak Price: ${data1['Close'].max():.2f}")
```

## Tasks 7, 8, 9: For Nike (January 1, 2022 to January 1, 2026)

- **Task 7**: Mean simple return (round to 6 dp) → **-0.000617**
- **Task 8**: Daily std dev of simple returns (round to 6 dp) → **0.023199**
- **Task 9**: Annual volatility of log returns (round to 6 dp) → **0.364036**

```python
prices1 = data1["Close"]
data1["Simple_Return"] = prices1.pct_change()
data1 = data1.dropna()  # Remove first row with NaN
data1["Log_Return"] = np.log(prices1 / prices1.shift(1))
data1 = data1.dropna()

print(f"NKE Mean Simple Return: {data1['Simple_Return'].mean():.6f}")
print(f"NKE Std Dev (Daily): {data1['Simple_Return'].std():.6f}")
print(f"NKE Volatility (Annual): {data1['Log_Return'].std() * np.sqrt(252):.6f}")
```

## Task 10: For Amazon, calculate the mean log return (from May 1, 2023 to May 1, 2025). Round to 5 decimal places.

**Answer: 0.00118**

```python
stock_amzn = yf.Ticker("AMZN")
data2 = stock_amzn.history(start="2023-05-01", end="2025-05-01", repair=True)

prices2 = data2["Close"]
data2["Log_Return"] = np.log(prices2 / prices2.shift(1))
data2 = data2.dropna()

print(f"AMZN Mean Log Return: {data2['Log_Return'].mean():.5f}")
```
